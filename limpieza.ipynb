```python
# ==========================================
# 1. SUBIR EL ARCHIVO A GOOGLE COLAB
# ==========================================
from google.colab import files
import pandas as pd
import numpy as np

print("Por favor, selecciona y sube tu archivo 'base de datos titanic.csv':")
uploaded = files.upload()

# Obtener el nombre del archivo subido automáticamente
nombre_archivo = list(uploaded.keys())[0]
df = pd.read_csv(nombre_archivo)

print("\n--- ARCHIVO CARGADO EXITOSAMENTE ---")
print(f"Filas originales: {df.shape[0]}, Columnas originales: {df.shape[1]}")

# ==========================================
# 2. LIMPIEZA Y TRANSFORMACIÓN DE DATOS
# ==========================================

# A. Eliminar columnas con exceso de nulos o alta cardinalidad no relevante
df_limpio = df.drop(columns=['Cabin', 'Ticket', 'PassengerId'])

# B. Imputar edad utilizando la mediana según la clase (Pclass) y el género (Sex)
df_limpio['Age'] = df_limpio.groupby(['Pclass', 'Sex'])['Age'].transform(lambda x: x.fillna(x.median()))

# C. Imputar los 2 valores nulos de la columna Embarked con la moda ('S')
df_limpio['Embarked'] = df_limpio['Embarked'].fillna(df_limpio['Embarked'].mode()[0])

# D. Convertir variables categóricas a numéricas
df_limpio['Sex_num'] = df_limpio['Sex'].map({'female': 1, 'male': 0})
df_limpio = pd.get_dummies(df_limpio, columns=['Embarked'], drop_first=True)

# E. Normalizar nombres de columnas a minúsculas
df_limpio.columns = df_limpio.columns.str.lower()

# ==========================================
# 3. VERIFICACIÓN Y DESCARGA AUTOMÁTICA
# ==========================================
print("\n--- RESUMEN DE VALORES NULOS TRAS LA LIMPIEZA ---")
print(df_limpio.isnull().sum())

# Guardar y descargar el archivo limpio
nombre_salida = 'titanic_limpio.csv'
df_limpio.to_csv(nombre_salida, index=False)
print(f"\n¡Limpieza completada! Descargando '{nombre_salida}'...")
files.download(nombre_salida)
```
